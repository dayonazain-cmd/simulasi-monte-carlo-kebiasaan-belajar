```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Generate data dummy yang mirip (skala 1-5, condong ke tengah-kanan)
np.random.seed(42)
# Menggabungkan dua distribusi agar bentuk kurva density-nya mirip asli
data_1 = np.random.normal(loc=2.5, scale=0.8, size=7000)
data_2 = np.random.normal(loc=3.5, scale=0.9, size=3000)
q12_data = np.concatenate([data_1, data_2])
q12_data = np.clip(q12_data, 1, 5)  # Batasi di skala 1-5

mean_val = 2.77  # Sesuai nilai di gambar Anda

# 2. Plotting Grafik Pertama
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Histogram dengan kurva KDE
sns.histplot(q12_data, bins=20, kde=True, color="#75baba", edgecolor="white", alpha=0.9)

# Kustomisasi Garis Mean
plt.axvline(mean_val, color="red", linestyle="--", linewidth=2, label=f"Mean: {mean_val}")

# Judul dan Label
plt.title("Distribusi Probabilitas Kepuasan Belajar (Q12)\n[Hasil 10.000 Iterasi Monte Carlo]", fontsize=12, pad=15)
plt.xlabel("Skala Kepuasan Belajar (1-5)", fontsize=11)
plt.ylabel("Frekuensi Muncul", fontsize=11)
plt.xlim(0.8, 5.2)
plt.legend(loc="upper right")

# Simpan sebagai gambar terpisah
plt.tight_layout()
plt.savefig("grafik_distribusi_q12.png", dpi=300)
plt.show()
```
